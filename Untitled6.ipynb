```python
# Colab cell (code)
import re
import os
import io
import joblib
from pathlib import Path
from tqdm.auto import tqdm
import pandas as pd
import numpy as np

# PDF and DOCX libraries
from PyPDF2 import PdfReader
import docx

# Function: clean text
def clean_text(text):
    if text is None:
        return ""
    # normalize whitespace and lower
    text = text.replace('\r', ' ').replace('\n', ' ')
    text = re.sub(r'http\S+', ' ', text)           # remove URLs
    text = re.sub(r'\w+@\w+\.\w+', ' ', text)      # remove emails
    text = re.sub(r'[^A-Za-z0-9\s\.\,]', ' ', text) # keep letters numbers spaces
    text = re.sub(r'\s+', ' ', text)               # collapse spaces
    text = text.strip().lower()
    return text

# Function: extract text from PDF
def extract_text_from_pdf(file_obj):
    try:
        reader = PdfReader(file_obj)
        text = []
        for p in reader.pages:
            page_text = p.extract_text()
            if page_text:
                text.append(page_text)
        return " ".join(text)
    except Exception as e:
        # Sometimes PyPDF2 fails for some PDFs
        return ""

# Function: extract text from DOCX
def extract_text_from_docx(file_obj):
    try:
        doc = docx.Document(file_obj)
        text = "\n".join([p.text for p in doc.paragraphs])
        return text
    except Exception as e:
        return ""

# Generic file->text dispatcher
def extract_text_from_file(uploaded_file):
    # uploaded_file may be a file-like object (like from gradio)
    name = getattr(uploaded_file, "name", None) or ""
    lower = name.lower()
    # If it's a bytes IO, we can wrap in BytesIO
    try:
        if lower.endswith(".pdf"):
            return extract_text_from_pdf(uploaded_file)
        elif lower.endswith(".docx") or lower.endswith(".doc"):
            return extract_text_from_docx(uploaded_file)
        elif lower.endswith(".txt") or lower.endswith(".csv"):
            # read as text
            try:
                uploaded_file.seek(0)
            except Exception:
                pass
            content = uploaded_file.read()
            if isinstance(content, bytes):
                try:
                    content = content.decode("utf-8")
                except:
                    content = content.decode("latin-1", errors="ignore")
            return content
        else:
            # fallback: try pdf then docx then read
            return extract_text_from_pdf(uploaded_file) or extract_text_from_docx(uploaded_file) or ""
    except Exception as e:
        return ""
```

```python
# Colab cell (code)
# Create a tiny sample dataset (for demo/training)
sample = [
    ("Experienced in Python, pandas, machine learning, data analysis and visualization.", "Data Science"),
    ("Developed web applications using Java, Spring Boot and REST APIs.", "Java Developer"),
    ("Network security engineer with experience firewalls, tcp/ip and penetration testing.", "Network Security"),
    ("Built responsive websites using HTML, CSS, JavaScript and React.", "Web Developer"),
    ("Experience in SAP, VLSI and embedded systems.", "Electronics Engineer"),
    ("Worked on blockchain and smart contracts using solidity and ethers.js", "Blockchain"),
    ("Managed logistics operations and supply chain tools.", "Operations Manager"),
    ("Professional fitness trainer and yoga coach.", "Health & Fitness"),
    ("Created automated test scripts using Selenium and pytest.", "Testing"),
    ("Worked on DevOps pipelines, Docker, Kubernetes and CI/CD.", "DevOps Engineer")
]

df_sample = pd.DataFrame(sample, columns=["resume_text","category"])
print("Sample dataset rows:", len(df_sample))
df_sample.head()
```

```python
# Colab cell (code)
# If you uploaded a CSV to /content, set its filename here:
csv_path = "/content/UpdatedResumeDataSet.csv"  # change if you uploaded with different name

if os.path.exists(csv_path):
    df = pd.read_csv(csv_path)
    # expecting columns like 'Resume' or 'resume_text' and 'Category' or 'category'
    if 'Resume' in df.columns and 'Category' in df.columns:
        df = df.rename(columns={'Resume':'resume_text','Category':'category'})
    elif 'resume' in df.columns and 'category' in df.columns:
        df = df.rename(columns={'resume':'resume_text','category':'category'})
    elif 'resume_text' in df.columns and 'category' in df.columns:
        pass
    else:
        print("CSV columns:", df.columns)
        raise ValueError("CSV doesn't contain expected columns. Please provide a file with resume_text and category.")
else:
    # use sample if CSV not present
    df = df_sample.copy()

# quick cleaning preview
df['cleaned'] = df['resume_text'].astype(str).apply(clean_text)
print("Data rows:", len(df))
display(df.head(10))
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.preprocessing import LabelEncoder
from sklearn.svm import LinearSVC
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, accuracy_score
import numpy as np
import joblib

X = df['cleaned'].values
y = df['category'].values

# encode labels
le = LabelEncoder()
y_enc = le.fit_transform(y)

# Determine if stratification is possible
y_enc_counts = np.bincount(y_enc)
# Check if all classes have at least 2 samples for stratification
stratify_condition = y_enc if np.all(y_enc_counts >= 2) else None

# split
X_train, X_test, y_train, y_test = train_test_split(X, y_enc, test_size=0.2, random_state=42, stratify=stratify_condition)

# vectorize
tfidf = TfidfVectorizer(stop_words='english', max_features=5000, ngram_range=(1,2))
X_train_tfidf = tfidf.fit_transform(X_train)
X_test_tfidf = tfidf.transform(X_test)

# train
clf = LinearSVC()
clf.fit(X_train_tfidf, y_train)

# evaluate
y_pred = clf.predict(X_test_tfidf)
print("Accuracy:", accuracy_score(y_test, y_pred))

# Get unique labels present in y_test and y_pred for classification report
unique_labels_in_test = np.unique(np.concatenate((y_test, y_pred)))
target_names_for_report = le.inverse_transform(unique_labels_in_test)

print(classification_report(y_test, y_pred, labels=unique_labels_in_test, target_names=target_names_for_report))

# Save artifacts for reuse
joblib.dump(tfidf, "/content/tfidf.pkl")
joblib.dump(le, "/content/label_encoder.pkl")
joblib.dump(clf, "/content/clf.pkl")
print("Saved tfidf.pkl, label_encoder.pkl, clf.pkl to /content")
```

```python
# Colab cell (code)
# Load artifacts (if you restarted runtime or running separate)
tfidf = joblib.load("/content/tfidf.pkl")
le = joblib.load("/content/label_encoder.pkl")
clf = joblib.load("/content/clf.pkl")

def predict_text(text):
    cleaned = clean_text(text)
    v = tfidf.transform([cleaned])
    pred = clf.predict(v)[0]
    return le.inverse_transform([pred])[0]

def predict_file_object(file_obj):
    # file_obj is a file-like object; ensure we can read from start
    try:
        file_obj.seek(0)
    except:
        pass
    raw_text = extract_text_from_file(file_obj)
    category = predict_text(raw_text)
    return category, raw_text
```

```python
# Colab cell (code)
import shutil

OUTPUT_DIR = Path("/content/output")
OUTPUT_DIR.mkdir(exist_ok=True, parents=True)

def segregate_and_save(uploaded_files):
    """
    uploaded_files: list of file-like objects (each should have .name)
    returns list of dicts: {filename, predicted_category, saved_path}
    """
    results = []
    for f in uploaded_files:
        fname = getattr(f, "name", None) or "uploaded_file"
        # ensure unique safe filename
        safe_name = Path(fname).name
        # run prediction
        try:
            f.seek(0)
        except:
            pass
        category, raw_text = predict_file_object(f)
        # create category folder
        cat_folder = OUTPUT_DIR / category.replace("/", "_")
        cat_folder.mkdir(parents=True, exist_ok=True)
        save_path = cat_folder / safe_name
        # Save the original uploaded bytes to file
        try:
            # file-like may be SpooledTemporaryFile or BytesIO
            with open(save_path, "wb") as out_f:
                try:
                    f.seek(0)
                except:
                    pass
                data = f.read()
                if isinstance(data, str):
                    data = data.encode("utf-8")
                out_f.write(data)
        except Exception as e:
            # fallback: write the extracted text into .txt file
            fallback_path = cat_folder / (safe_name + ".txt")
            with open(fallback_path, "w", encoding="utf-8") as out_f:
                out_f.write(raw_text)
            save_path = fallback_path
        results.append({"filename": safe_name, "predicted_category": category, "saved_path": str(save_path)})
    return results
```

```python
# Colab cell (code)
import gradio as gr

def predict_single_text_ui(text):
    if not text or text.strip()=="":
        return "Please paste resume text or upload a file"
    cat = predict_text(text)
    return cat

def bulk_predict_ui(files):
    """
    files: list of uploaded file-like objects (gradio)
    """
    if not files:
        return "Upload one or more resume files (pdf/docx/txt)"
    results = segregate_and_save(files)
    # create a simple dataframe for display
    df_res = pd.DataFrame(results)
    return df_res  # Gradio will render dataframe

with gr.Blocks() as demo:
    gr.Markdown("# Resume Screening App (Bulk + Single)\n- Paste text to predict single resume\n- Or upload multiple files (pdf/docx/txt) to bulk predict and save into /content/output/<category>/")
    with gr.Row():
        with gr.Column(scale=1):
            txt_in = gr.Textbox(label="Paste Resume Text", lines=6, placeholder="Paste resume text here...")
            txt_out = gr.Label(label="Predicted Category for Text")
            btn_txt = gr.Button("Predict Text")
            btn_txt.click(fn=predict_single_text_ui, inputs=txt_in, outputs=txt_out)
        with gr.Column(scale=1):
            files_in = gr.File(label="Bulk Upload (multiple resumes) - pdf/docx/txt", file_count="multiple")
            out_table = gr.Dataframe(headers=["filename","predicted_category","saved_path"], label="Bulk Prediction Results")
            btn_files = gr.Button("Run Bulk Prediction + Segregate")
            btn_files.click(fn=bulk_predict_ui, inputs=files_in, outputs=out_table)

    gr.Markdown("Output files saved into /content/output/<category>/ (you can download from Colab file browser or use !zip to download).")

# Launch with share True to get a public link
demo.launch(share=True)
```

```python
# ===============================
# 1) INSTALL DEPENDENCIES
# ===============================
!pip install patool
!pip install python-docx
!pip install PyPDF2
!pip install gradio

import os
import patoolib
import zipfile
import shutil
from PyPDF2 import PdfReader
from docx import Document
import gradio as gr

# ===============================
# 2) PATH TO YOUR UPLOADED FILE
# ===============================
input_path = "/content/predicted_outputs.zip"  # <-- CHANGE NAME IF NEEDED
output_base = "/content/resume_dataset"
temp_extract_path = "/content/temp_extracted_resumes" # Temporary directory for initial extraction

# Clean old folders
if os.path.exists(output_base):
    shutil.rmtree(output_base)
os.makedirs(output_base, exist_ok=True)

if os.path.exists(temp_extract_path):
    shutil.rmtree(temp_extract_path)
os.makedirs(temp_extract_path, exist_ok=True)

# ===============================
# 3) EXTRACT ZIP / RAR FILES
# ===============================
def extract_all(input_path, output_path):
    if input_path.endswith(".zip"):
        with zipfile.ZipFile(input_path, 'r') as z:
            z.extractall(output_path)
    elif input_path.endswith(".rar"):
        patoolib.extract_archive(input_path, outdir=output_path)
    else:
        print("Unsupported file type.")

extract_all(input_path, temp_extract_path) # Extract to the temporary directory

# ===============================
# 4) READ FILE CONTENT (PDF, DOCX, TXT)
# ===============================
def read_resume(file_path):
    text = ""

    if file_path.endswith(".txt"):
        text = open(file_path, "r", errors="ignore").read()

    elif file_path.endswith(".docx"):
        doc = Document(file_path)
        for p in doc.paragraphs:
            text += p.text + "\n"

    elif file_path.endswith(".pdf"):
        try:
            reader = PdfReader(file_path)
            for page in reader.pages:
                text += page.extract_text() or ""
        except:
            text = ""
    return text.lower()

# ===============================
# 5) CATEGORY AUTOMATIC DETECTION
# ===============================
categories = {
    "Data Science": ["python", "machine learning", "pandas", "numpy", "tensorflow", "kaggle"],
    "Java Developer": ["java", "spring", "hibernate", "j2ee"],
    "Network Security": ["network", "security", "cyber", "firewall", "vpn"],
    "Testing": ["selenium", "test cases", "automation", "qa"],
    "Web Developer": ["html", "css", "javascript", "react", "node"]
}

# Create folders in the final output_base
for cat in categories:
    os.makedirs(f"{output_base}/{cat}", exist_ok=True)
os.makedirs(f"{output_base}/Others", exist_ok=True)

# ===============================
# 6) CLASSIFY RESUMES INTO CATEGORIES
# ===============================
def classify_resume(text):
    for cat, keywords in categories.items():
        if any(word in text for word in keywords):
            return cat
    return "Others"

# Process each file from the temporary extraction path
for root, dirs, files in os.walk(temp_extract_path): # Walk through the temporary directory
    for file in files:
        file_path = os.path.join(root, file)
        if file.endswith((".pdf", ".docx", ".txt")):
            content = read_resume(file_path)
            category = classify_resume(content)

            # Ensure the target category directory exists in output_base
            os.makedirs(f"{output_base}/{category}", exist_ok=True)
            shutil.copy(file_path, f"{output_base}/{category}/{file}")

# Clean up the temporary extracted directory
if os.path.exists(temp_extract_path):
    shutil.rmtree(temp_extract_path)

# ===============================
# 7) GET CATEGORY LIST
# ===============================
def get_categories():
    return [d for d in os.listdir(output_base) if os.path.isdir(os.path.join(output_base, d))]

# ===============================
# 8) GET FILES INSIDE CATEGORY
# ===============================
def get_files(category):
    cat_path = f"{output_base}/{category}"
    if not os.path.exists(cat_path):
        return []
    return os.listdir(cat_path)

# ===============================
# 9) SEARCH INSIDE RESUMES
# ===============================
def search_resumes(keyword):
    keyword = keyword.lower()
    results = []

    for cat in get_categories():
        for f in get_files(cat):
            fp = f"{output_base}/{cat}/{f}"
            text = read_resume(fp)
            if keyword in text:
                results.append(f"{f}  →  ({cat})")

    return results

# ===============================
# 10) GRADIO UI
# ===============================
def ui_list_files(category):
    return "\n".join(get_files(category))

def ui_search(keyword):
    res = search_resumes(keyword)
    return "\n".join(res) if res else "No resumes found."

with gr.Blocks() as app:
    gr.Markdown("# │ Resume Category Browser + Search")

    category = gr.Dropdown(get_categories(), label="Select Category")
    show_btn = gr.Button("Show Resumes")
    output1 = gr.Textbox(label="Files in Category")

    show_btn.click(ui_list_files, category, output1)

    gr.Markdown("### │ Search Inside All Resumes")
    key = gr.Textbox(label="Enter skill / keyword")
    search_btn = gr.Button("Search")
    output2 = gr.Textbox(label="Search Results")

    search_btn.click(ui_search, key, output2)

app.launch()
```
